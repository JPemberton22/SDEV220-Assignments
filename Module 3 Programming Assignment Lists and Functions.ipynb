```python
# 9.2
def get_odds():
    for n in range(10):
        if n % 2 == 1:
            yield n

counter = 0
third_value = None
for val in get_odds():
    counter += 1
    if counter == 3:
        third_value = val
        break

print("third odd from range(10) is", third_value)
```
