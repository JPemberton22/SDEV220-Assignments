```python
# Sort 0s, 1s, and 2s
def sort012(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:
        if arr[mid] == 0:
            # Swap arr[low] and arr[mid]
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1

        elif arr[mid] == 1:
            mid += 1

        else:
            # Swap arr[mid] and arr[high]
            arr[high], arr[mid] = arr[mid], arr[high]
            high -= 1

    return arr
```

```python
# Binary Search
def binary_search_first(arr, k):
    low = 0
    high = len(arr) - 1
    result = -1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == k:
            result = mid
            high = mid - 1

        elif arr[mid] < k:
            low = mid + 1
        else:
            high = mid - 1

    return result
```
